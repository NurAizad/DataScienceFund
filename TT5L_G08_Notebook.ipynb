```python
#FINDING OUTLIERS
def findingOutliers(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1

    lowerLimit = Q1 - 1.5*IQR
    upperLimit = Q3 + 1.5*IQR
    return (column < lowerLimit) | (column > upperLimit)
```

#### Elbow method for Clustering

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
ss = StandardScaler()
cluster_scaled = ss.fit_transform(accidents_cluster)
print(accidents_cluster)
print(cluster_scaled)
```

```python
# Code adapted from W3Schools (n.d)
#https://www.w3schools.com/python/python_ml_k-means.asp
inertia =[]
for i in range(1,11):
    kmeans = KMeans(n_clusters=i, random_state = 0)
    kmeans.fit(cluster_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(range(1,11), inertia, marker='o')
plt.title('Elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.show()
```

6 is chosen as the number of clusters, because it is the first point where the drop starts to flatten

```python
model = KMeans(6, verbose=0, n_init=10,random_state = 0)
model.fit(cluster_scaled)
```

```python
accidents_cluster['cluster'] = model.labels_
```

```python
v1 = accidents_cluster['crash_hour']
v2 = accidents_cluster['injuries_total']

fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111)
scatter = ax.scatter(v1, v2, c=accidents_cluster['cluster'],s=50,cmap='jet',alpha=0.7)
ax.set_title('K-Means Clustering')
ax.set_xlabel(v1.name)
ax.set_ylabel(v2.name)
plt.colorbar(scatter)
plt.show()
```

According to the evaluation report for classification, random forest has higher accuracy as compared to decision tree. Both models have higher accuracy to predict Class 2 which is damage that is over $1500 because of its greater number as compared to Class 0 and Class 1. 

For the confusion matrix, same conclusion can be constructed as Class 2 has higher number as compared to Class 0 and Class 1. So it can guesses Class 2 more. Therefore, enhancement must be conducted using SMOTE.


### Classification Enhancement using SMOTE

```python
# Code adapted from GeeksforGeeks and Imbalanced Learn
#https://www.geeksforgeeks.org/machine-learning/smote-for-imbalanced-classification-with-python/
#https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SMOTE.html

from imblearn.over_sampling import SMOTE

smote = SMOTE(sampling_strategy='not majority', random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)
```

```python
dec_tree_smote = DecisionTreeClassifier(random_state=42)
rand_forest_smote = RandomForestClassifier(random_state=42)

# Train models
dec_tree_smote.fit(X_train_smote, y_train_smote)
rand_forest_smote.fit(X_train_smote, y_train_smote)
```

```python
plt.figure(figsize=(10,10))
tree.plot_tree(dec_tree_smote, max_depth=4, fontsize = 12)
```

### Classification Using SMOTE Evaluation

```python
models = {'Decision Tree': dec_tree_smote, 'Random Forest': rand_forest_smote}

for name, model in models.items():
    y_pred = model.predict(X_test)
    print(f"{name}:")
    print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
    print(classification_report(y_test, y_pred))
```

```python
for name, model in models.items():
    y_pred = model.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(5,4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f'Confusion Matrix - {name}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()
```

After applying SMOTE, the accuracy to predict Class 0 and 1 are improved. Although the overall accuracy decreases for both decision tree and random forest, it can predict Class 0 and 1 more accurately.

From the confusion matrix, after applying SMOTE, the number of correctly guessed Class 0 and Class 1 increase from 2035 to 2828 and 1260 to 6838 respectively for decision tree. While for random forest, Class 0 increase from 1980 to 2846 and Class 1 increases from 1045 to 6719. Class 1 has a significant improvement in the prediction.
